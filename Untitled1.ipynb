```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

# Parámetros ILUSTRATIVOS
m1, m2 = 1.5, 1.0  # kg
l1, l2 = 0.70, 0.60  # m
lc1, lc2 = l1/2, l2/2  # m
I1, I2 = m1*l1**2/12, m2*l2**2/12  # kg m²; barras uniformes
g = 9.81  # m/s²
b1, b2 = 0.12, 0.08  # N m s/rad
fc1, fc2 = 0.08, 0.06  # N m
EPS = 0.02  # rad/s; regularización suave de Coulomb

def M(q):
    q1, q2 = q
    c = np.cos(q2)
    return np.array([
        [m1*lc1**2 + I1 + m2*(l1**2 + lc2**2 + 2*l1*lc2*c) + I2,
         m2*(lc2**2 + l1*lc2*c) + I2],
        [m2*(lc2**2 + l1*lc2*c) + I2, m2*lc2**2 + I2]
    ])

def C(q, dq):
    h = m2*l1*lc2*np.sin(q[1])
    dq1, dq2 = dq
    # Convención equivalente a la matriz de la página 282 del libro
    return np.array([[-2*h*dq2, -h*dq2], [h*dq1, 0.0]])

def G(q):
    q1, q2 = q
    return g*np.array([
        (m1*lc1 + m2*l1)*np.sin(q1) + m2*lc2*np.sin(q1+q2),
        m2*lc2*np.sin(q1+q2)
    ])

def F(dq):
    # Modelo suave viscoso + Coulomb para permitir integración numérica.
    # La fricción estática exacta del libro exige una lógica adicional de adhesión.
    return np.array([b1*dq[0]+fc1*np.tanh(dq[0]/EPS),
                     b2*dq[1]+fc2*np.tanh(dq[1]/EPS)])

def q_des(t):
    return np.array([0.35*np.sin(0.7*t), 0.25*np.sin(0.9*t)])

def dq_des(t):
    return np.array([0.245*np.cos(0.7*t), 0.225*np.cos(0.9*t)])

def f(t, x):
    q, dq = x[:2], x[2:]
    tau = G(q) + np.array([8., 5.])*(q_des(t)-q) + np.array([3.,2.])*(dq_des(t)-dq)
    ddq = np.linalg.solve(M(q), tau - C(q,dq)@dq - G(q) - F(dq))
    return np.r_[dq,ddq]

t = np.linspace(0, 10, 501)
sol = solve_ivp(f, (0,10), [0.15,-0.10,0.,0.], t_eval=t, rtol=1e-8, atol=1e-10)
assert sol.success, sol.message
plt.figure(figsize=(10,4))
for i in (0,1):
    plt.plot(t,np.rad2deg(sol.y[i]),label=f'q{i+1} simulada')
    plt.plot(t,np.rad2deg([q_des(z)[i] for z in t]),'--',label=f'q{i+1} deseada')
plt.xlabel('Tiempo (s)'); plt.ylabel('Ángulo (grados)')
plt.title('Posiciones articulares — robot de 2 GDL (10 s)')
plt.grid(alpha=.3); plt.legend(); plt.show()

def points(q):
    p0=np.array([0.,0.]); q1,q2=q
    p1=p0 + l1*np.array([np.sin(q1),-np.cos(q1)])
    p2=p1 + l2*np.array([np.sin(q1+q2),-np.cos(q1+q2)])
    return np.array([p0,p1,p2])

fig, ax = plt.subplots(figsize=(6,6))
ax.set_aspect('equal'); ax.set_xlim(-1.4,1.4); ax.set_ylim(-1.5,.5)
ax.grid(alpha=.25)
line, = ax.plot([],[],'o-',lw=5,markersize=9)
time_txt = ax.text(.02,.95,'',transform=ax.transAxes)

def update(i):
    p = points(sol.y[:2,i]); line.set_data(p[:,0],p[:,1]);
    time_txt.set_text(f't = {t[i]:.2f} s')
    return line,time_txt

anim=FuncAnimation(fig,update,frames=range(0,len(t),5),interval=80,blit=True)
plt.close(fig)
display(HTML(anim.to_jshtml()))

```
